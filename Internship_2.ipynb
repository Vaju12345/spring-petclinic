```python
df["Demand_level"]=pd.cut(df["MonthlyOrders"],bins=[0,500,2000,5000],labels=["Low","medium","high"])
df
```

```python
df["Growth_category"]=df["GrowthFactor"].apply(lambda x:"High Growth" if x>1.2 else "stable" if x>=0.9 else "declining")
df
```

```python
df["AOV_level"]=pd.cut(df["AOV"],bins=[0,500,2000,5000],labels=["low","medium","high"])
df
```

```python
df["Growth_Score"] = (
    df["MonthlyOrders"] * 0.4 +
    df["GrowthFactor"] * 1000 * 0.3 +
    df["AOV"] * 0.3
)
```

```python
df
```

```python
df["Total_Share"] = df["InStoreShare"] + df["UE_share"] + df["DD_share"] + df["SD_share"]
df
```

```python
df["Dominant_Channel"] = df[["InStoreShare", "UE_share", "DD_share", "SD_share"]].idxmax(axis=1)
```

```python
df["Channel_Diversity"] = 1 - (
    df[["InStoreShare", "UE_share", "DD_share", "SD_share"]]
    .max(axis=1)
)
df
```

```python
df["Online_Share"] = df["UE_share"] + df["DD_share"] + df["SD_share"]
df["Offline_Share"] = df["InStoreShare"]
```

```python
df["Reach_Category"] = pd.cut(df["DeliveryRadiusKM"],
                             bins=[0, 3, 7, 15],
                             labels=["Local", "Medium", "Wide"])
df
```
